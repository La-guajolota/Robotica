```markdown
### Entrenamiento de un modelo de red neuronal convolucional

En esta celda, se creó y entrenó un modelo de red neuronal convolucional utilizando TensorFlow y Keras. A continuación, se describen los pasos realizados:

1. **Importación de módulos**: Se importaron las bibliotecas necesarias de TensorFlow y Keras para construir y entrenar el modelo.

2. **Definición del modelo**:
    - Se utilizó un modelo secuencial (`Sequential`) con las siguientes capas:
      - Una capa convolucional 2D (`Conv2D`) con 32 filtros, un tamaño de kernel de 3x3, función de activación ReLU y una forma de entrada de `(64, 64, 3)`.
      - Una capa de aplanamiento (`Flatten`) para convertir los datos en un vector unidimensional.
      - Una capa densa (`Dense`) con una neurona y función de activación sigmoide para la salida.

3. **Compilación del modelo**:
    - Se utilizó el optimizador Adam.
    - La función de pérdida fue `binary_crossentropy`.

4. **Generación de datos aleatorios**:
    - Se generaron datos de entrada `x` con forma `(64, 64, 64, 3)` y valores aleatorios.
    - Se generaron etiquetas `y` con 64 valores aleatorios binarios (0 o 1).

5. **Entrenamiento del modelo**:
    - El modelo fue entrenado durante 2 épocas con un tamaño de lote (`batch_size`) de 16.

Este proceso permitió entrenar un modelo básico de clasificación binaria utilizando datos generados aleatoriamente.
```

````python
import tensorflow as tf
from tensorflow.keras import layers, models

# Modelo trivial
model = models.Sequential([
    layers.Conv2D(32, 3, activation='relu', input_shape=(64,64,3)),
    layers.Flatten(),
    layers.Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy')

# Datos aleatorios
x = tf.random.normal([64,64,64,3])
y = tf.random.uniform([64], maxval=2, dtype=tf.int32)

# Entrena 2 épocas
model.fit(x, y, epochs=2, batch_size=16)
````
